```python
import numpy as np

# Create two NumPy arrays
array1 = np.array([10, 20, 30, 40, 50])
array2 = np.array([2, 4, 5, 8, 10])

print("Array 1:", array1)
print("Array 2:", array2)

# Check array shapes
print("Shape of Array 1:", array1.shape)
print("Shape of Array 2:", array2.shape)
```

```python
# Addition
addition = array1 + array2

# Subtraction
subtraction = array1 - array2

# Multiplication
multiplication = array1 * array2

# Division
division = array1 / array2

# Display results
print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)
```

### Vectorized Operations in NumPy

Vectorization is a technique in NumPy that allows mathematical operations to be performed directly on entire arrays without using explicit loops. It applies operations element-wise, making the code shorter, simpler, and generally faster than using traditional Python loops.

In this task, vectorized operations are used to perform addition, subtraction, multiplication, and division on two NumPy arrays. Since both arrays have the same shape, the operations are performed element-wise.
